```python
import pandas as pd
import numpy as np
import os

df = pd.read_csv('../data/raw/cs-training.csv')
print(df.shape)
df.head()
```

### Changing Columns Name

```python
df = df.rename(columns={
    "Unnamed: 0": "ID",
    "SeriousDlqin2yrs": "Default",
    "RevolvingUtilizationOfUnsecuredLines": "CreditUtilization",
    "age": "Age",
    "NumberOfTime30-59DaysPastDueNotWorse": "Late30-59DaysPastDue",
    "DebtRatio": "DebtRatio",
    "MonthlyIncome": "MonthlyIncome",
    "NumberOfOpenCreditLinesAndLoans": "OpenCreditLines",
    "NumberOfTimes90DaysLate": "Late90DaysPastDue",
    "NumberRealEstateLoansOrLines": "RealEstateLoans",
    "NumberOfTime60-89DaysPastDueNotWorse": "Late60-89DaysPastDue",
    "NumberOfDependents": "Dependents"
})

df.columns.tolist()
```

### Checking Duplicate Rows

```python
print("Duplicate Rows:", df.duplicated().sum())
print("Duplicate IDs:", df['ID'].duplicated().sum())
```

### Adjusting Age 0 value

```python
df[df['Age'] <= 0]
df.loc[df['Age'] <= 0, 'Age'] = np.nan
```

### Checking the Missing Summary

```python
missing_summary = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percentage": df.isnull().mean() * 100
})

missing_summary[missing_summary['missing_count'] > 0]
```

### Checking Negative Values

```python
numeric_column = [
    col for col in df.columns
    if col not in ['ID', 'Default']
]

for col in numeric_column:
    negative_count = (df[col] < 0).sum()

    if negative_count > 0:
        print(col, "->", negative_count, "negative values")
```

### Creating Clean Feature List

```python
TARGET = "Default"
ID = "ID"

FEATURES = [
    col for col in df.columns
    if col not in [TARGET, ID]
]

print("Number of Features:", len(FEATURES))
print(FEATURES)
```

### Saving the Cleaned Data

```python
os.makedirs('../data/processed', exist_ok=True)

df.to_csv('../data/processed/clean_credit.csv', index=False)

print("Cleaned Dataset Saved...")
```
